```python
import duckdb
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = duckdb.sql("SELECT * FROM 'data/software-data.parquet'").df()
print(df.shape)
df.head()
```

```python
df = df.ffill()
pd.set_option("display.max_columns", None)
print(df.head(100).to_string())
```

```python
df["t"] = df["Time"] - df["Time"].iloc[0]

df["v_ms"] = df["SME_TRQSPD_Speed"] * (12/41) * (2*np.pi/60) * 0.2
df["v_mph"] = df["v_ms"] * 2.23694

plt.figure(figsize=(10, 4))
plt.plot(df["t"], df["v_ms"], label="Motor speed")
plt.plot(df["t"], df["VDM_GPS_SPEED"], label="GPS speed", alpha=0.6)
plt.xlabel("Time (s)")
plt.ylabel("Speed (mph)")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 4))
plt.plot(df["t"], df["ETC_STATUS_BRAKE_SENSE_VOLTAGE"], label="Brake Voltage")
plt.xlabel("Time (s)")
plt.ylabel("Brake Power (V)")
plt.legend()
plt.show()
```

```python
moving = df["v_ms"] > 0.5

df["braking"] = (df["ETC_STATUS_BRAKELIGHT"] == 1) & moving
df["accelerating"] = (df["SME_THROTL_TorqueDemand"] > 0) & ~df["braking"] & moving
df["coasting"] = ~df["braking"] & ~df["accelerating"] & moving

dt = df["t"].diff().mean()
print(df[["accelerating", "braking", "coasting"]].sum() * dt)
```

```python
def get_segments(mask):
    group = (mask != mask.shift()).cumsum()
    seg = df[mask].groupby(group[mask])["t"].agg(["min", "max"])
    seg["duration"] = seg["max"] - seg["min"]
    return seg.reset_index(drop=True)

acc_segs = get_segments(df["accelerating"])
brk_segs = get_segments(df["braking"])
coast_segs = get_segments(df["coasting"])

print(acc_segs.head(10))
print(brk_segs.head(10))
print(coast_segs.head(10))
```

```python
fig, axes = plt.subplots(5, 1, figsize=(12, 8), sharex=True)

moving = (df["v_mph"] > 0.5).astype(int)
braking = (df["ETC_STATUS_BRAKE_SENSE_VOLTAGE"] > 352)
accelerating = (df["SME_THROTL_TorqueDemand"] > 0)
coasting = ((moving) & (~braking) & (~accelerating))

axes[0].plot(df["t"], df["v_mph"])
axes[0].set_ylabel("Speed (mph)")

axes[1].plot(df["t"], (accelerating).astype(int))
axes[1].set_ylabel("Accelerating")

axes[2].plot(df["t"], (braking).astype(int))
axes[2].set_ylabel("Braking") #340-345 noise, above 350 is likely braking

axes[3].plot(df["t"], (coasting).astype(int))   
axes[3].set_ylabel("Coasting")

axes[4].plot(df["t"], df["ETC_STATUS_BRAKE_SENSE_VOLTAGE"])
axes[4].set_ylabel("Brake sense V") #346-348 is without braking
axes[4].set_ylim(336, 374)

axes[-1].set_xlabel("Time (s)") 
plt.tight_layout()
plt.show()

```

```python
lap_start = (0.0031, 0.00045)
dfunf = duckdb.sql("SELECT * FROM 'data/software-data.parquet'").df()
good = (
    dfunf["VDM_GPS_Latitude"].notna() & dfunf["VDM_GPS_Longitude"].notna()
    & (dfunf["VDM_GPS_Latitude"] != 0) & (dfunf["VDM_GPS_Longitude"] != 0)
)
good_gps = dfunf.loc[good, ["Time", "VDM_GPS_Latitude", "VDM_GPS_Longitude"]].copy()
good_gps["x_m"] = (good_gps["VDM_GPS_Longitude"] - good_gps["VDM_GPS_Longitude"].iloc[0]) * 111320 * np.cos(np.radians(good_gps["VDM_GPS_Latitude"].iloc[0]))
good_gps["y_m"] = (good_gps["VDM_GPS_Latitude"] - good_gps["VDM_GPS_Latitude"].iloc[0]) * 110540


print(good_gps)
```

```python
x0 = -5.685 

good_gps["t"] = good_gps["Time"] - good_gps["Time"].iloc[0]
xs = good_gps["x_m"].values
ts = good_gps["t"].values

crossings = 0
entry_times = []
passed_east = False
passed_west = False

for i in range(1, len(xs)):
    if xs[i] > 100:
        passed_east = True
    if xs[i] < -15:
        passed_west = True

    crossed = (xs[i-1] <= x0) and (xs[i] > x0)

    if crossed:
        first_entry = (crossings == 0)
        if first_entry or (passed_east and passed_west):
            frac = (x0 - xs[i-1]) / (xs[i] - xs[i-1])
            t_cross = ts[i-1] + frac * (ts[i] - ts[i-1])

            crossings += 1
            entry_times.append(t_cross)
            passed_east = False
            passed_west = False

print("Laps:", crossings - 1)
print("Entry times:", np.round(entry_times, 2))
```

```python
m_eff = 244.08
m = 221.4
g = 9.81
rho = 1.225
```

```python
coasting = moving & ~braking & ~accelerating
mask = coasting & (df["v_ms"] >= 5)
print("Rows that are coasting above 5 m/s:", mask.sum())
```

```python
seg_id = (mask != mask.shift()).cumsum()
```

```python
good_segments = []

for sid, seg in df[mask].groupby(seg_id[mask]):
    duration = seg["t"].iloc[-1] - seg["t"].iloc[0]
    if duration >= 1.0:
        seg = seg.copy()
        seg["seg"] = sid
        good_segments.append(seg)

print("Number of coastdowns kept:", len(good_segments))

cd = pd.concat(good_segments)
```

```python
win = 25   # smoothing window in rows, about 0.3 s

v_list = []
a_list = []

for sid, seg in cd.groupby("seg"):
    v = seg["v_ms"].rolling(win, center=True, min_periods=1).mean().values
    t = seg["t"].values
    a = np.gradient(v, t)
    v_list.append(v)
    a_list.append(a)

v = np.concatenate(v_list)
a = np.concatenate(a_list)

print("Points for the fit:", len(v))
print("Average deceleration (m/s^2):", a.mean())
```

```python
x = v**2
slope, intercept = np.polyfit(x, a, 1)

print("slope:", slope)
print("intercept:", intercept)
```

```python
CdA = -slope * m_eff / (0.5 * rho)
Crr = -intercept * m_eff / (m * g)

print("Cd*A (m^2):", CdA)
print("Crr:", Crr)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, a, s=4, alpha=0.3, label="Coastdown data")

xs = np.linspace(x.min(), x.max(), 100)
plt.plot(xs, slope * xs + intercept, "r", label="Linear fit")

plt.xlabel("v² (m²/s²)")
plt.ylabel("dv/dt (m/s²)")
plt.legend()
plt.show()
```
